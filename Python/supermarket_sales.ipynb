```python
# Objective :
# The objective of this project is to analyze supermarket sales data to identify sales trends, customer behavior, product performance, and payment perferences. Theanalysis helps business managers make data_driven decisions to improve sales and customer satisfaction.

# Dataset Description :
#The dataset contains sales transactions from a supermarket. It includes information about branches, cities,customer type ,product lines, sales amounts, payment methods, and customer ratings.
```

```python
#import libraries

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load dataset

df = pd.read_csv("supermarket_sales.csv")

print(df.head())
```

```python
# Understand the Dataset

print(df.info())
```

```python
print(df.describe())
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
# Data Cleaning

# remove extra space from column names

df.columns = df.columns.str.strip()
```

```python
# check duplicates 

print(df.duplicated().sum())
```

```python
#check missing values

print(df.isnull().sum())
```

```python
#what are the total sale
print("Total Sales =", df["Total"].sum())
```

```python
# what is the average sale

print("Average Sales =",df["Total"].mean())
```

```python
# highest sale 

print(df["Total"].max())
```

```python
#Lowest sale

print(df["Total"].min())
```

```python
#Branch wise sale

branch_sale = df.groupby("Branch")["Total"].sum()

print(branch_sale)
```

```python
#Product_wise sales

product_sales = df.groupby("Product_Line")["Total"].sum()

print(product_sales)
```

```python
# Payment Method Analysis

payment = df["Payment"].value_counts()

print(payment)
```

```python
#Customer type Analysis

customer = df.groupby("Customer_Type")["Total"].sum()

print(customer)
```

```python
#City wis Sales

city_sales = df.groupby("City")["Total"].sum()

print(city_sales)
```

```python
#Average rating by product line

rating =df.groupby("Product_Line")["Rating"].mean()

print(rating)
```

```python
#visualization

branch_sales.plot(kind="bar")

plt.title("Branch_wise Sales")
plt.xlabel("Branch")
plt.ylabel("Sales")

plt.show()
```

```python
#product wise sale

product_sales.plot(kind="bar")

plt.title("Product-wise Sales")
plt.xlabel("product")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

```python
#Payment Method

payment.plot(kind="pie", autopct="%1.1f%%")

plt.title("Payment Method Distribution")
plt.ylabel("")

plt.show()
```

```python
#City wise sales

city_sales.plot(kind="bar")

plt.title("City_wise Sales")

plt.show()
```

```python
# Business Insights 
#1. Branch B generated the highest sales.
#2. Electronics was the highest revenue-generating product line.
#3. Cash and card were the most preferred payment methods.
#4. Member customerscontributed significantly to sales.
#5. Hyderabad generated strong revenu compared to other cities.
```

```python
#Recomendations
#1. Increase inventory for high_selling product.
#2.Improve promotions in low_performing branches.
#3.Introduce loyalty offers to increase member purchases.
#4. Encourage digital payment methods with cashback offers.
#5. Focus marketing camaigns on top_performing product categories.
```

```python
#Best performing city

print(df.groupby("City")["Total"].sum().sort_values(ascending=False))
```

```python
#customer type distribution

df["Customer_Type"].value_counts().plot(
    kind="bar"
)

plt.title("Customer Type Distribution")
plt.xlabel("Customer Type")
plt.ylabel("Count")

plt.show()
```

```python
#Rating Distribution

plt.hist(df["Rating"])

plt.title("Customer Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")

plt.show()
```

```python
#Sales by Payment Method 

payment_sales = df.groupby("Payment")["Total"].sum()

payment_sales.plot(kind="bar")

plt.title("Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Sales")

plt.show()
```

```python
# Conclusion

# The analysis shows that sales performance varies by branch, product line,and city.Electronics generated the highest revenue,while Branch B contributed the most to total sales.Customer ratings remained consistently high, indicating good service quality.
#These insight can help management improve inventory planing, marketing campaigns ,and customer engagement strategies.
```
